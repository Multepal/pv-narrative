# Ajtzibab: Parse (sentence-level)

OHCO: `para_num, sent_num, token_num`

Source lines from the existing `ajtzibab-DOC.csv` and `ajtzibab-DOCMAP.csv` (line_id level);
lines serve as sentence proxies for this poetic edition.

```python
import pandas as pd
import re
```

```python
src_id = 'ajtzibab'
base = '../../notebooks'
```

## Load existing line-level DOC and DOCMAP

```python
DOC_old    = pd.read_csv(f'{base}/ajtzibab/ajtzibab-DOC.csv',    index_col='doc_id')
DOCMAP_old = pd.read_csv(f'{base}/ajtzibab/ajtzibab-DOCMAP.csv', index_col='doc_id')
```

## Rename line_id → sent_num (0-indexed)

```python
DOCMAP = DOCMAP_old.rename(columns={'line_id': 'sent_num'}).copy()
DOCMAP['sent_num'] = DOCMAP['sent_num'] - 1
DOC = DOC_old.copy()
print(f'{len(DOC):,} sentences from {DOCMAP.para_num.nunique()} paragraphs')
DOCMAP.head()
```

## DOC to TOKEN

```python
TOKEN = DOC.doc_str.str.split(expand=True).stack().to_frame('token_str')
TOKEN.index.names = DOC.index.names + ['token_num']
TOKEN['term_str'] = TOKEN.token_str.str.lower().str.replace(r"[^a-z']", '', regex=True)
TOKEN = TOKEN[TOKEN.term_str != ''].dropna()
TOKEN
```

## Save

```python
TOKEN.to_csv(f'{src_id}-TOKEN.csv')
DOC.to_csv(f'{src_id}-DOC.csv')
DOCMAP.to_csv(f'{src_id}-DOCMAP.csv')
print('Saved to notebooks/doc_tables/')
```
