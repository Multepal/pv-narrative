```python
import pandas as pd
```

```python
chunk_files = !ls *CHUNK*
chunk_dfs = []
src_ids = []
for cf in chunk_files:
    src_id = cf.split('-')[0]
    src_ids.append(src_id)
    df = pd.read_csv(cf).set_index('chunk_num')
    chunk_dfs.append(df)
```

```python
CHUNKMAP = pd.concat(chunk_dfs, axis=1, keys=src_ids)
CHUNKMAP.T.swaplevel().sort_index().T['chunk_str']
```

```python
CHUNKMAP.to_csv("CHUNKMAP.csv", index=True)
```
