```python
import pandas as pd
```

```python
!ls *CHUNK*
```

```python
n_chunks = 60
chunk_files = !ls *CHUNK-{n_chunks}*
chunk_str_dfs = []
chunk_doc_dfs = []
src_ids = []
for cf in chunk_files:
    src_id = cf.split('-')[0]
    src_ids.append(src_id)
    df = pd.read_csv(cf).set_index('chunk_num')
    chunk_str_dfs.append(df.chunk_str)
    chunk_doc_dfs.append(df.doc_id)
```

```python
CHUNKSTR = pd.concat(chunk_str_dfs, axis=1, keys=src_ids)
CHUNKSTR.head()
```

```python
CHUNKSTR.to_csv("CHUNKSTR.csv", index=True)
```

```python
CHUNKDOC = pd.concat(chunk_doc_dfs, axis=1, keys=src_ids)
CHUNKDOC.columns.name = 'src_id'
CHUNKDOC.head()
```

```python
# CHUNKSTR.colop
```

```python
CHUNKDOC2 = CHUNKDOC.stack().to_frame('doc_ids').doc_ids.str.split(expand=True).stack().dropna().to_frame('doc_id')
CHUNKDOC2.index.names = ['chunk_num', 'src_id', 'doc_num']
CHUNKDOC2
```

```python
E = CHUNKDOC2.groupby('src_id')
E_colop = E.get_group('colop')
E_colop
# E_colop.doc_id.value_counts().sort_index()
```
